### Q4 — Mean Squared Error (MSE)
MSE is the average of the squared differences between actual and predicted values. Example: actual values `[3, 5, 7]` and predictions `[2, 5, 9]` give squared errors `[1, 0, 4]`, so MSE = `(1 + 0 + 4) / 3 = 1.667`. Squaring penalizes large errors more strongly.

### Q5 — Sigmoid function
Logistic regression converts a linear score `z` into a probability using `sigmoid(z) = 1 / (1 + exp(-z))`. For `z = 0`, the probability is `0.5`; positive scores yield probabilities above `0.5`, negative scores below `0.5`.

### Q9 — K-Fold Cross-Validation
Divide the training data into `k` folds. Train `k` times, holding out a different fold for validation each time, then average the validation scores. For five folds, each fold is used once for validation and four times as part of training. Keep a separate final test set untouched.

### Q15 — End-to-end ML workflow
Define the problem and target; collect and inspect data; split data; fit preprocessing on training data only; choose candidate models; train and tune using validation or cross-validation; evaluate once on the held-out test set; document, deploy, and monitor as appropriate. Avoid leakage and selecting hyperparameters using the test set.
